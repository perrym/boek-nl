## Doel

Ontbrekende waarden, duplicaten en datatypes automatisch samenvatten.

## Opdracht

Maak de voorbeelddata aan of lees een eigen CSV in. Bouw een rapport met datatype, aantal ontbrekende waarden, percentage ontbrekend en aantal unieke waarden per kolom.



Werk stap voor stap:

3. Voer de gevraagde selectie, opschoning of aggregatie uit.

5. Beschrijf welke datakwaliteitsfout de analyse het sterkst kan beïnvloeden.

Hint: Bewaar de ruwe data apart en voer bewerkingen uit op een kopie wanneer je wilt kunnen terugkijken.

## Werkwijze

1. Probeer de opdracht eerst zelf.
2. Voer de code uit en controleer de tussenresultaten.
3. Voeg minimaal één randgeval toe.
4. Open pas daarna `uitwerking.py` en `uitwerking.md`.


```
import pandas as pd

gegevens = pd.DataFrame({
    "incident_id": [1, 2, 2, 4],
    "categorie": ["IAM", "Netwerk", "Netwerk", None],
    "doorlooptijd": [4.5, 8.0, None, 2.0],
})

rapport = pd.DataFrame({
    "datatype": gegevens.dtypes.astype(str),
    "ontbrekend": gegevens.isna().sum(),
    "percentage_ontbrekend": gegevens.isna().mean().mul(100).round(1),
    "uniek": gegevens.nunique(dropna=True),
})

print(rapport)
print("Dubbele rijen:", gegevens.duplicated().sum())
```

# Uitwerking 6.1

## Hoe werkt de code?

Een datakwaliteitsrapport is een eerste technische controle. Het percentage ontbrekend maakt kolommen vergelijkbaar wanneer datasets verschillend groot zijn. Let op dat dubbele ID’s niet per definitie volledige dubbele rijen zijn. Daarvoor is een aparte businessregel nodig.


## Verwachte uitvoer

De uitvoer meldt een ontbrekende categorie, een ontbrekende doorlooptijd en drie unieke incident-ID’s ondanks vier rijen.


## Controleer je uitkomst

- Ontbrekende waarden worden per kolom geteld.
- ID-uniciteit wordt apart beoordeeld.
- Datatypes worden zichtbaar gemaakt.

## Verdiepingsopdracht

Schrijf het rapport naar datakwaliteit.csv en voeg minimum en maximum voor numerieke kolommen toe.
