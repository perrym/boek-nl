## Opdracht

Maak een kleine DataFrame met een kolom doorlooptijd en een kolom datum waarin bewust één ongeldige waarde staat. Zet beide kolommen gecontroleerd om en rapporteer welke waarden niet konden worden geconverteerd.

De ongeldige tekstwaarden worden niet stilzwijgend als geldige data behandeld. Door de NaN- en NaT-resultaten expliciet te tellen, wordt zichtbaar welke bronwaarden nader onderzoek nodig hebben.

Van tabellen naar visualisatie

Met Pandas kunnen we nu tabulaire data gecontroleerd inlezen, inspecteren, filteren, groeperen en opschonen. Daarmee hebben we de stap gemaakt van losse arrays naar datasets met betekenisvolle kolomnamen, categorieën en datatypes.

De volgende stap is resultaten begrijpelijk presenteren. Een tabel is geschikt voor detail, maar patronen, verdelingen en afwijkingen zijn vaak sneller zichtbaar in een grafiek. In het volgende hoofdstuk gebruiken we daarom de opgeschoonde DataFrames als basis voor visualisaties. Daarbij blijft dezelfde regel gelden: een grafiek is alleen betrouwbaar wanneer de onderliggende selectie, populatie en datakwaliteit duidelijk zijn.
