```
import pandas as pd

gegevens = pd.DataFrame({
    "categorie": ["IAM", "Netwerk", "IAM", None, "Applicatie", "IAM"]
})

categorie = gegevens["categorie"].fillna("Onbekend")
samenvatting = categorie.value_counts().rename_axis("categorie").reset_index(name="aantal")
samenvatting["percentage"] = (samenvatting["aantal"] / len(gegevens) * 100).round(1)

print(samenvatting)
```

# Uitwerking 6.2

## Hoe werkt de code?

fillna voorkomt dat ontbrekende waarden stil uit de telling verdwijnen. value_counts sorteert standaard van hoog naar laag. Door het resultaat naar een DataFrame om te zetten kun je het later exporteren of visualiseren.


## Verwachte uitvoer

IAM heeft drie van de zes records en krijgt 50,0 procent.


## Controleer je uitkomst

- De totalen tellen op tot het aantal rijen.
- Ontbrekende categorieën zijn zichtbaar.
- Percentages tellen ongeveer op tot 100 procent.
