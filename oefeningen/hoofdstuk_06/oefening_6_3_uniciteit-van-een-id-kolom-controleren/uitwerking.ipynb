## Opdracht

Controleer of incident_id uniek en niet leeg is. Toon alle rijen met een dubbele of ontbrekende ID en laat het script met een foutcode stoppen wanneer de controle faalt.

4. Open pas daarna `uitwerking.py` en `uitwerking.md`.


```
import pandas as pd


gegevens = pd.DataFrame({
    "incident_id": [1001, 1002, 1002, None],
    "omschrijving": ["MFA", "Firewall", "Firewall kopie", "Onbekend"],
})

problemen = gegevens[
    gegevens["incident_id"].isna()
    | gegevens["incident_id"].duplicated(keep=False)
]

if problemen.empty:
    print("Controle geslaagd: alle ID's zijn uniek en gevuld")
    exitcode = 0
else:
    print("Controle mislukt:")
    print(problemen.to_string(index=False))
    exitcode = 1

print(f"Verwachte script-exitcode: {exitcode}")
assert exitcode == 1
```

**Notebook-opmerking:** de bijbehorende `uitwerking.py` stopt bewust met exitcode 1 wanneer de controle faalt, zoals in het boek beschreven. In dit notebook wordt die exitcode als waarde gecontroleerd, zodat **Restart Kernel and Run All** volledig kan afronden.

# Uitwerking 6.3

## Hoe werkt de code?

duplicated(keep=False) markeert alle voorkomens van een dubbele waarde. Daardoor zie je niet alleen de tweede rij, maar ook de eerste rij waarmee deze botst. Een niet-nul exitcode is bruikbaar in een pipeline, omdat de verwerking dan automatisch kan worden geblokkeerd.


## Verwachte uitvoer

Het script toont beide rijen met ID 1002 en de rij zonder ID. Daarna eindigt het met exitcode 1.


## Controleer je uitkomst

- Alle voorkomens van duplicaten worden getoond.
- Lege ID’s gelden als fout.
- De controle kan een pipeline laten falen.
