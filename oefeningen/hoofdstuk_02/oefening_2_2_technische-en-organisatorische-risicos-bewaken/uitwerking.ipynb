```
import pandas as pd

risicos = [
    {"fase": "Data", "type": "Technisch", "risico": "Verouderde documenten", "kans": 4, "impact": 4},
    {"fase": "Data", "type": "Organisatorisch", "risico": "Geen data-eigenaar", "kans": 3, "impact": 5},
    {"fase": "Ontwikkeling", "type": "Technisch", "risico": "Prompt injection", "kans": 4, "impact": 5},
    {"fase": "Productie", "type": "Organisatorisch", "risico": "Geen incidentproces", "kans": 2, "impact": 5},
]

df = pd.DataFrame(risicos)
df["score"] = df["kans"] * df["impact"]
df = df.sort_values("score", ascending=False)
print(df.to_string(index=False))
```

# Uitwerking 2.2

## Hoe werkt de code?

Kans maal impact is een eenvoudige prioritering, geen volledig risicomodel. Het helpt wel om bespreekbaar te maken welke risico's eerst moeten worden behandeld. Een technische maatregel alleen is niet genoeg wanneer eigenaarschap, acceptatiecriteria of incidentafhandeling ontbreken.


## Verwachte uitvoer

Prompt injection krijgt in dit voorbeeld score 20 en staat bovenaan.


## Controleer je uitkomst

- Kans en impact gebruiken dezelfde schaal.
- De score wordt niet verward met een objectieve waarheid.
- Voor hoge risico's wordt een eigenaar en maatregel toegevoegd.

## Verdiepingsopdracht
