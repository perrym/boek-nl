# Oefening 4.1 - risicoscores classificeren

**Niveau:** Basis    Geschatte tijd: 15 tot 20 minuten

## Doel

Oefenen met een list, een for-lus en een if-voorwaarde.

## Opdracht

Maak een lijst met de scores 25, 60, 85, 40 en 
90. Doorloop de lijst en toon per score of deze Hoog (70 of hoger) of Laag of middel is. Tel daarna hoeveel hoge scores voorkomen.

## Werkwijze

1. Probeer de opdracht eerst zelf.
2. Voer de code uit en controleer de tussenresultaten.
3. Voeg minimaal één randgeval toe.
4. Open pas daarna `uitwerking.py` en `uitwerking.md`.


```
scores = [25, 60, 85, 40, 90]
aantal_hoog = 0

for score in scores:
    if score >= 70:
        print(score, "Hoog")
        aantal_hoog += 1
    else:
        print(score, "Laag of middel")

print("Aantal hoge scores:", aantal_hoog)
```
