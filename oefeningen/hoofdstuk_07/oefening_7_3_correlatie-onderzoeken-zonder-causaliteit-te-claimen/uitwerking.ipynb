```
import matplotlib.pyplot as plt
import numpy as np

rng = np.random.default_rng(7)
wijzigingen = rng.integers(5, 80, size=60)
incidenten = 2 + 0.18 * wijzigingen + rng.normal(0, 3, size=60)

correlatie = np.corrcoef(wijzigingen, incidenten)[0, 1]
print(f"Correlatie: {correlatie:.2f}")
print("Verklaring 1: grotere systemen hebben zowel meer wijzigingen als meer incidenten.")
print("Verklaring 2: slechte change-controls veroorzaken een deel van beide waarden.")

plt.scatter(wijzigingen, incidenten)
plt.xlabel("Aantal wijzigingen")
plt.ylabel("Aantal incidenten")
plt.title("Wijzigingen en incidenten")
plt.tight_layout()
plt.show()
```

# Uitwerking 7.3

## Hoe werkt de code?

Een correlatie beschrijft samenhang, maar bewijst geen oorzaak. Een derde variabele, zoals systeemgrootte, kan beide waarden beïnvloeden. Ook kan de relatie andersom lopen of door meetfouten ontstaan. De code combineert daarom een statistiek met expliciete alternatieve verklaringen.


## Verwachte uitvoer

De precieze correlatie is door de vaste random seed reproduceerbaar en positief.


## Controleer je uitkomst

- De grafiek wordt visueel gecontroleerd.
- Er wordt geen causale conclusie getrokken.
- Mogelijke confounders, oftewel verstorende factoren, worden benoemd.
