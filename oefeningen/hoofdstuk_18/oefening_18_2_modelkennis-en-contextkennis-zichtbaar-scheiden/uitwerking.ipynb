## Doel

Voorkomen dat een taalmodel onbekende actuele feiten verzint.

## Opdracht

Bouw een eenvoudige antwoordfunctie die alleen feitelijke informatie uit meegeleverde context gebruikt. Wanneer het antwoord niet in de context staat, moet de functie dat expliciet melden.



Werk stap voor stap:


1. Schrijf eerst het doel en het gewenste uitvoerformaat op.

2. Maak een kleine invoer met voldoende context en één bewust ontbrekend gegeven.

3. Bouw de prompt, kostenberekening of beslismatrix in Python.

4. Controleer of de uitvoer compleet, gestructureerd en controleerbaar is.

5. Test daarna een onduidelijke invoer en beschrijf hoe het systeem daarmee omgaat.

Hint: Beperk de oefening tot één duidelijk doel. Een langere prompt is niet automatisch een betere prompt.

## Werkwijze

1. Probeer de opdracht eerst zelf.
2. Voer de code uit en controleer de tussenresultaten.
3. Voeg minimaal één randgeval toe.
4. Open pas daarna `uitwerking.py` en `uitwerking.md`.


```
context = {
    "beleid_versie": "2026.3",
    "mfa_verplicht": True,
    "eigenaar": "Identity & Access Management",
}


def beantwoord(vraag: str, context: dict) -> str:
    vraag_lager = vraag.lower()
    if "versie" in vraag_lager:
        return f"De beleidsversie is {context['beleid_versie']}."
    if "mfa" in vraag_lager:
        return "MFA is verplicht." if context["mfa_verplicht"] else "MFA is niet verplicht."
    if "eigenaar" in vraag_lager:
        return f"De eigenaar is {context['eigenaar']}."
    return "Niet te beantwoorden op basis van de beschikbare context."

for vraag in ["Welke versie geldt?", "Is MFA verplicht?", "Wanneer is het beleid goedgekeurd?"]:
    print(vraag, "->", beantwoord(vraag, context))
```

# Uitwerking 18.2

## Hoe werkt de code?

Dit is geen echt LLM, maar de oefening maakt het verschil scherp. Modelkennis is wat een model uit training heeft geleerd. Contextkennis wordt tijdens de vraag meegegeven en kan actueel, controleerbaar en brongebonden zijn. Een professioneel systeem moet onbekende informatie niet aanvullen met een geloofwaardige gok.


## Verwachte uitvoer

De eerste twee vragen worden beantwoord. De goedkeuringsdatum resulteert in een expliciete onbekendmelding.


## Controleer je uitkomst

- Ontbrekende context wordt niet verzonnen.
- Context heeft een eigenaar en versie.
- Antwoorden kunnen naar hun bron worden herleid.

## Verdiepingsopdracht

Laat de functie ook een bronveld en zekerheidsniveau retourneren.


## Zelf verder testen

Verander minimaal één invoerwaarde en voeg een randgeval toe. Controleer daarna of de conclusie nog steeds door de uitvoer wordt ondersteund.
