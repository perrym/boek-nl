```
import pandas as pd

werkzaamheden = [
    {"taak": "Doorlooptijd van een project voorspellen", "categorie": "regressie"},
    {"taak": "Phishingmail herkennen", "categorie": "classificatie"},
    {"taak": "Vergelijkbare incidenten groeperen", "categorie": "clustering"},
    {"taak": "Conceptbevinding schrijven", "categorie": "generatieve AI"},
    {"taak": "Factuurbedrag optellen", "categorie": "geen AI"},
    {"taak": "Risicoklasse van een API bepalen", "categorie": "classificatie"},
]

df = pd.DataFrame(werkzaamheden)
print(df)
print("\nAantal per categorie:")
print(df["categorie"].value_counts())
```

# Uitwerking 1.1

## Hoe werkt de code?

Een AI-project begint bij het probleem en niet bij het model. De tabel dwingt je om per taak vast te stellen welk soort uitvoer nodig is. Een getal voorspellen is regressie, een label kiezen is classificatie en nieuwe tekst maken valt onder generatieve AI. De taak "factuurbedrag optellen" is bewust geen AI, omdat gewone code daar betrouwbaarder en goedkoper is.


## Verwachte uitvoer

De uitvoer toont zes werkzaamheden. Classificatie komt twee keer voor en de overige categorieën ieder eenmaal. De code maakt een overzichtstabel van verschillende taken en hun bijbehorende AI-categorie. Vervolgens telt Python hoe vaak iedere categorie voorkomt. Zo zie je snel welke soorten taken in deze voorbeeldlijst het meest voorkomen.

De categorie past bij het soort uitvoer.

Ook niet-AI-oplossingen worden overwogen.

De tabel kan zonder handmatige telling worden samengevat.
