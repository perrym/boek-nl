# Oefening 26.1 - Een risicomatrix voor agenttools berekenen

**Niveau:** Verdieping  
**Geschatte tijd:** 45 tot 75 minuten  
**Benodigde packages:** pandas

## Doel

Toolrechten, data-impact en omkeerbaarheid vertalen naar een risicoscore.

## Opdracht

Maak vijf tools en beoordeel per tool de leesrechten, schrijfrechten, gevoeligheid en omkeerbaarheid. Bereken vervolgens de totaalscore en bepaal de classificatie.



Werk stap voor stap:

Opdracht: Risicobeoordeling van AI-tools

Maak vijf voorbeeldtools.

Leg per tool de leesrechten, schrijfrechten, datagevoeligheid en omkeerbaarheid vast.

Definieer de scoreschaal.

Gebruik voor ieder criterium een consistente schaal en beschrijf duidelijk wat elke waarde betekent.

Bereken de totaalscore.

Tel de scores samen en koppel de uitkomst aan een risicoklasse, bijvoorbeeld laag, middel of hoog.

Controleer uitzonderingen.

Controleer of een tool met schrijfrechten op gevoelige data niet door een lage totaalscore ten onrechte als laag risico wordt geclassificeerd.

Bepaal beheersmaatregelen.

Voeg voor tools met een hoog risico passende maatregelen toe, zoals menselijke goedkeuring, aanvullende autorisatie of een beperktere scope.

Hint: Een somscore is alleen triage. Harde veiligheidsvoorwaarden, zoals ongeautoriseerde schrijfacties, moeten de score kunnen overrulen.
