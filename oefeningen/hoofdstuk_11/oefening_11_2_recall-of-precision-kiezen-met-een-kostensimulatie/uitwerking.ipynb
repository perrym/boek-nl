```
modellen = {
    "Model Hoog-Recall": {"tp": 95, "fp": 80, "fn": 5},
    "Model Hoog-Precision": {"tp": 75, "fp": 15, "fn": 25},
}

kosten_fp = 40
kosten_fn = 5000

for naam, m in modellen.items():
    precision = m["tp"] / (m["tp"] + m["fp"])
    recall = m["tp"] / (m["tp"] + m["fn"])
    totale_kosten = m["fp"] * kosten_fp + m["fn"] * kosten_fn
    print(
        f"{naam}: precision={precision:.2%}, recall={recall:.2%}, "
        f"kosten=EUR {totale_kosten:,}"
    )
```

# Uitwerking 11.2

## Hoe werkt de code?

Bij kritieke kwetsbaarheden kan een gemiste bevinding veel duurder zijn dan een extra handmatige beoordeling. Het hoge-recallmodel veroorzaakt meer false positives, maar mist veel minder kritieke gevallen. De gekozen kosten zijn aannames en moeten met proceseigenaren worden gevalideerd.


## Verwachte uitvoer

Met deze kosten is het hoge-recallmodel veel goedkoper, ondanks de lagere precision.


## Controleer je uitkomst

- Foutkosten zijn traceerbaar naar een aanname of bron.
- Capaciteit voor handmatige review wordt meegenomen.
- De gekozen threshold wordt gemonitord.
