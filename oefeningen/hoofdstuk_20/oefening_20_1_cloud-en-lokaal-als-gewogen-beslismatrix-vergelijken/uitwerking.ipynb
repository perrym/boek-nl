## Opdracht

Maak een DataFrame met criteria, gewichten en scores voor cloud en lokaal. Bereken de gewogen totaalscore.



Werk stap voor stap:

2. Geef ieder criterium een gewicht en leg kort vast waarom dat gewicht past bij de use-case.

3. Ken cloud en lokaal per criterium een score toe en bereken met Pandas de gewogen scores.

4. Vergelijk de totaalscores en controleer of de gekozen schaal voor beide opties hetzelfde is.

5. Verander minimaal één belangrijk gewicht en beoordeel of de conclusie gevoelig is voor die keuze.

```
import pandas as pd

criteria = pd.DataFrame([
    {"criterium": "Dataprivacy", "gewicht": 5, "cloud": 3, "lokaal": 5},
    {"criterium": "Schaalbaarheid", "gewicht": 4, "cloud": 5, "lokaal": 3},
    {"criterium": "Beheerlast", "gewicht": 3, "cloud": 5, "lokaal": 2},
    {"criterium": "Offline beschikbaar", "gewicht": 2, "cloud": 1, "lokaal": 5},
    {"criterium": "Modelkwaliteit", "gewicht": 5, "cloud": 5, "lokaal": 3},
])

for optie in ["cloud", "lokaal"]:
    criteria[f"gewogen_{optie}"] = criteria["gewicht"] * criteria[optie]

print(criteria.to_string(index=False))
print("Cloud totaal:", criteria["gewogen_cloud"].sum())
print("Lokaal totaal:", criteria["gewogen_lokaal"].sum())
```

# Uitwerking 20.1

## Hoe werkt de code?

Een beslismatrix maakt aannames zichtbaar. De uitkomst is alleen zo goed als de gewichten en scores. Privacy kan bijvoorbeeld een harde eis zijn in plaats van een compensabel criterium. Dan mag een hoge schaalbaarheid een onvoldoende privacycontrole niet wegstrepen.


## Verwachte uitvoer

De tabel toont per criterium de gewogen bijdrage en een totaalscore voor beide opties.


## Controleer je uitkomst

- Harde eisen worden apart gemarkeerd.
- Scores zijn onderbouwd met meetgegevens.
- De matrix wordt bij gewijzigde omstandigheden opnieuw uitgevoerd.

## Verdiepingsopdracht

Voeg een gevoeligheidsanalyse toe waarin ieder gewicht 1 punt hoger en lager wordt getest.
