**Niveau:** Basis

## Doel

Vastleggen welke Python-, platform- en packageversies voor een project zijn gebruikt.

## Opdracht

Maak of activeer een aparte omgeving, installeer uitsluitend de benodigde packages en voer daarna het script uit.

Maak of activeer een aparte Python-omgeving.

Installeer alleen NumPy en Pandas.

Voer het voorbeeld uit en noteer versies, platform en eventuele foutmeldingen.

Herhaal de uitvoering vanuit een schone terminal.

Leg in twee zinnen uit waarom de uitkomst reproduceerbaar is.

Hint

Een foutmelding over een ontbrekend package is nuttige informatie: zij toont dat de omgeving werkelijk schoon is.

## Werkwijze

1. Probeer de opdracht eerst zelf.
2. Voer de code uit en controleer de tussenresultaten.
3. Voeg minimaal één randgeval toe.
4. Open pas daarna `uitwerking.py` en `uitwerking.md`.


```
import platform

import sys

import numpy as np

import pandas as pd

assert sys.version_info >= (3, 10), "Python 3.10 of hoger is vereist"

print("Python:", sys.version.split()[0])

print("Platform:", platform.platform())

print("NumPy:", np.__version__)

print("Pandas:", pd.__version__)
```

# Uitwerking 3.1

## Hoe werkt de code?

Het script toont de gebruikte Python-versie, het platform en de geïnstalleerde versies van NumPy en Pandas. De assert stopt met een duidelijke melding wanneer Python te oud is. De uitvoer kan als technisch bewijs bij een test of release worden bewaard.


## Verwachte uitvoer

Je ziet vier herkenbare versieregels. Bij Python ouder dan 3.10 stopt het script met een foutmelding.


## Controleer je uitkomst

- De omgeving heeft een herkenbare projectnaam.
- Dependencies zijn geëxporteerd.
- De versie-informatie wordt bij een release bewaard.
