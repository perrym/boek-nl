```
import pandas as pd

vergelijking = pd.DataFrame([
    {
        "model": "CNN",
        "lokale_patronen": 5,
        "lange_afstandsrelaties": 2,
        "databehoefte": 3,
        "rekencapaciteit": 3,
        "typische_toepassing": "beeldclassificatie",
    },
    {
        "model": "Transformer",
        "lokale_patronen": 3,
        "lange_afstandsrelaties": 5,
        "databehoefte": 5,
        "rekencapaciteit": 5,
        "typische_toepassing": "tekst en multimodale modellen",
    },
])

print(vergelijking.to_string(index=False))
print("\nGemiddelde technische zwaarte:")
print(vergelijking.set_index("model")[["databehoefte", "rekencapaciteit"]].mean(axis=1))
```

# Uitwerking 17.1

## Hoe werkt de code?

De scores zijn geen universele feiten, maar een expliciete beoordelingsschaal. Een CNN verwerkt lokale patronen efficiënt en is vaak een sterke keuze voor beperkte beelddata. Transformers modelleren relaties over grotere afstanden en zijn zeer flexibel, maar kunnen meer data en rekencapaciteit vragen.


## Verwachte uitvoer

De tabel maakt de afweging zichtbaar en berekent een eenvoudige gemiddelde technische zwaarte.


## Controleer je uitkomst

- De schaal en betekenis van scores zijn gedocumenteerd.
- De keuze hangt af van dataset en taak.
- Een eenvoudiger baseline wordt niet overgeslagen.
