# Oefening 22.2 - Chunk grootte van 200 en 1200 woorden vergelijken

**Niveau:** Gemiddeld tot gevorderd  
**Geschatte tijd:** 40 tot 60 minuten

## Doel

Het effect van kleine en grote chunks op context, aantal stukken en zoekprecisie onderzoeken.

## Opdracht

Maak een chunk functie op basis van woorden. Verdeel dezelfde beleidstekst in chunks van 200 en 1200 woorden met overlap. Vergelijk aantal chunks en gemiddelde lengte.



Werk stap voor stap:


1. Gebruik hetzelfde document voor beide chunkgroottes zodat de vergelijking eerlijk blijft.

2. Maak chunks van ongeveer 200 en 1200 woorden en leg overlap en bron-ID vast.

3. Voer dezelfde set zoekvragen uit tegen beide varianten.

4. Vergelijk hoeveel relevante context wordt gevonden en hoeveel irrelevante tekst mee wordt opgehaald.

5. Beschrijf welk compromis je ziet tussen context, precisie, kosten en traceerbaarheid.

Hint: Er bestaat geen universeel beste chunkgrootte. Kies op basis van documentstructuur en meetbare retrievalkwaliteit.

## Werkwijze

1. Probeer de opdracht eerst zelf.
2. Voer de code uit en controleer de tussenresultaten.
3. Voeg minimaal één randgeval toe.
4. Open pas daarna `uitwerking.py` en `uitwerking.md`.


```
def chunk_woorden(tekst: str, grootte: int, overlap: int) -> list[str]:
    if overlap >= grootte:
        raise ValueError("Overlap moet kleiner zijn dan chunkgrootte")
    woorden = tekst.split()
    stap = grootte - overlap
    return [" ".join(woorden[i:i + grootte]) for i in range(0, len(woorden), stap)]

# Simuleer een beleidsdocument met herkenbare secties.
secties = [
    "Doel en reikwijdte van het informatiebeveiligingsbeleid.",
    "Identiteitsbeheer vereist MFA en periodieke toegangsreviews.",
    "Incidenten moeten binnen vastgestelde termijnen worden gemeld.",
    "Leveranciers worden beoordeeld op beveiliging en continuïteit.",
]
tekst = " ".join(secties * 400)

for grootte, overlap in [(200, 40), (1200, 120)]:
    chunks = chunk_woorden(tekst, grootte, overlap)
    gemiddelde = sum(len(c.split()) for c in chunks) / len(chunks)
    print(f"grootte={grootte}, overlap={overlap}, aantal={len(chunks)}, gemiddeld={gemiddelde:.1f}")
```

# Uitwerking 22.2

## Hoe werkt de code?

Kleine chunks geven meestal specifiekere zoekresultaten, maar kunnen context over sectiegrenzen verliezen. Grote chunks bewaren meer samenhang, maar voegen mogelijk irrelevante tekst toe en kosten meer tokens. De beste keuze hangt af van documentstructuur, vraagtype en embeddingmodel.


## Verwachte uitvoer

De 200-woordinstelling produceert veel meer chunks dan de 1200-woordinstelling.


## Controleer je uitkomst

- Overlap is kleiner dan de chunkgrootte.
- Koppen en tabellen worden waar mogelijk structureel behandeld.
- De vergelijking gebruikt echte evaluatievragen.
