# Oefening 22.1 - Metadata voor interne documenten ontwerpen en valideren

**Niveau:** Gemiddeld tot gevorderd  
**Geschatte tijd:** 40 tot 60 minuten

## Doel

Zoekfilters, bronverwijzingen en toegangscontrole ondersteunen.

## Opdracht

Maak meta datarecords voor interne documenten. Schrijf een validatiefunctie die verplichte velden, toegestane classificaties en datumformaat controleert.



Werk stap voor stap:


1. Bepaal welke metadata nodig is voor autorisatie, actualiteit en herleidbaarheid, bijvoorbeeld document-ID, eigenaar, classificatie, status en versie.

2. Maak enkele voorbeeldrecords met bewust één ontbrekend of ongeldig veld.

3. Valideer verplichte velden en toegestane waarden voordat documenten worden geïndexeerd.

4. Toon welke records worden geaccepteerd en welke moeten worden geblokkeerd of hersteld.

5. Leg uit welke metadata als hard retrievalfilter moet gelden en welke alleen voor ranking of rapportage wordt gebruikt.

Hint: Autorisatie- en geldigheids metadata zijn geen vrijblijvende zoekbonus; behandel ze waar nodig als harde voorwaarden.

## Werkwijze

1. Probeer de opdracht eerst zelf.
2. Voer de code uit en controleer de tussenresultaten.
3. Voeg minimaal één randgeval toe.
4. Open pas daarna `uitwerking.py` en `uitwerking.md`.


```
from datetime import date

VERPLICHT = {"document_id", "titel", "eigenaar", "classificatie", "versie", "geldig_vanaf"}
CLASSIFICATIES = {"Intern", "Vertrouwelijk", "Openbaar"}


def valideer_metadata(metadata: dict) -> list[str]:
    fouten = []
    ontbrekend = VERPLICHT - metadata.keys()
    if ontbrekend:
        fouten.append(f"Ontbrekende velden: {sorted(ontbrekend)}")
    if metadata.get("classificatie") not in CLASSIFICATIES:
        fouten.append("Ongeldige classificatie")
    try:
        date.fromisoformat(metadata.get("geldig_vanaf", ""))
    except ValueError:
        fouten.append("geldig_vanaf moet YYYY-MM-DD zijn")
    return fouten

record = {
    "document_id": "DOC-2026-014",
    "titel": "Identity Access Review",
    "eigenaar": "Kennisbeheer",
    "classificatie": "Vertrouwelijk",
    "versie": "1.2",
    "geldig_vanaf": "2026-05-01",
    "toegestane_rollen": ["reviewer", "review_manager"],
}

print(valideer_metadata(record))
```

# Uitwerking 22.1

## Hoe werkt de code?

Metadata is niet alleen zoekinformatie. Het ondersteunt versiebeheer, bronverwijzingen, filters en autorisatie. De toegestane_rollen moeten voor retrieval worden gecontroleerd, zodat een taalmodel nooit context ontvangt die de gebruiker niet mag zien.


## Verwachte uitvoer

Een leeg foutenoverzicht betekent dat het record aan de basisregels voldoet.


## Controleer je uitkomst

- Document-ID is uniek.
- Versie en geldigheidsdatum zijn aanwezig.
- Autorisatie wordt niet alleen in de prompt geregeld.

## Verdiepingsopdracht

Voeg een vervaldatum en hash van de documentinhoud toe.
