# Oefening 32.2 - Twintig acceptatietests als Python-testcatalogus maken

**Niveau:** Verdieping  
**Geschatte tijd:** 45 tot 75 minuten  
**Benodigde packages:** pandas

## Doel

Functionele, kwaliteits-, performance- en securityeisen aantoonbaar maken.

## Opdracht

Maak twintig testcases, waaronder minimaal vijf securitytests. Controleer aantallen en verplichte velden.



Werk stap voor stap:


1. Definieer testcategorieën voor functionaliteit, kwaliteit, performance, betrouwbaarheid en security.

2. Maak twintig testcases en zorg dat iedere testcase een ID, categorie, verwacht resultaat en eigenaar heeft.

3. Controleer automatisch dat minimaal vijf securitytests aanwezig zijn en dat verplichte velden nergens ontbreken.

4. Voeg minimaal één test voor onvoldoende bewijs, één autorisatietest en één prompt-injection test toe.

5. Genereer een samenvatting per categorie en rapporteer welke acceptatie-eisen nog niet aantoonbaar zijn.

Hint: Een testcatalogus is pas bewijs wanneer het verwachte resultaat vooraf is vastgelegd en de uitvoering reproduceerbaar is.
